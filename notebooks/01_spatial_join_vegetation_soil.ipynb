```python
import geopandas as gpd
```

```python
file3 = "/content/vegselctionervalider.shx"
file1 = "/content/adminSelectioneer.shx"
file2 = "/content/soilselectioner.shx"
file4 = "/content/fireselec.shp"


gdf1 = gpd.read_file(file1)
gdf4 = gpd.read_file(file4)
gdf2 = gpd.read_file(file2)
```

```python
# Effectuer l'intersection spatiale
intersection = gpd.overlay(gdf1, gdf2, how='intersection')

# Enregistrer le résultat dans un nouveau fichier shapefile
#output_file = "/content/adminsoil.shp"
#intersection.to_file(output_file)
```

```python
gdf3 = gpd.read_file(file3)
```

```python
intersection = gpd.overlay(intersection, gdf3, how='intersection')
```

```python
jointure = gpd.sjoin(intersection,gdf4, how='left', op='intersects')
```

```python
colonnes_a_conserver = ['sand % topsoil', 'silt % topsoil', 'clay % topsoil', 'pH water topsoil', 'OC % topsoil',
                        'N % topsoil', 'BS % topsoil', 'CEC topsoil', 'CEC clay topsoil', 'CaCO3 % topsoil',
                        'BD topsoil', 'C/N topsoil', 'ID', 'GRIDCODE', 'AREA', 'LCCCODE', 'fid', 'admin_level',
                        'osm_admin_level', 'name', 'name_en', 'population', 'hasc', 'LATITUDE', 'LONGITUDE',
                        'BRIGHTNESS', 'SCAN', 'TRACK', 'ACQ_DATE', 'ACQ_TIME', 'SATELLITE', 'INSTRUMENT',
                        'CONFIDENCE', 'VERSION', 'BRIGHT_T31', 'FRP', 'DAYNIGHT', 'TYPE']

# Sauvegarder le résultat au format CSV
output_csv = "/content/resultat_jointurepy.csv"
jointure[colonnes_a_conserver].to_csv(output_csv, index=False)
```
